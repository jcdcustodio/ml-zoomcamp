```python
def normalize_name(s: str) -> str:
    # Insert underscore between consecutive capitals followed by lowercase (e.g., 'XMLHttp' -> 'XML_Http')
    s = re.sub(r"([A-Z]+)([A-Z][a-z])", r"\1_\2", s)
    # Insert underscore between lowercase/numbers and following uppercase (e.g., 'HttpReq' -> 'Http_Req')
    s = re.sub(r"([a-z\d])([A-Z])", r"\1_\2", s)

    return s.lower()
```

```python
# Apply to every categorical column and sort the results from most to least importance

def mutual_info_churn_score(feature):
    return mutual_info_score(feature, df_full_train["churn"])

mis = df_full_train[categorical_features].apply(mutual_info_churn_score)
mis.sort_values(ascending=False)
```

Essentially, mutual information is a measure of the importance of a categorical variable (how much we learn about the target if we have the information from a particular feature)

## One-hot encoding

```python
from sklearn.feature_extraction import DictVectorizer
```

```python
df[categorical_features + numerical_features].to_dict(orient="records")[:5]
```

```python
dv = DictVectorizer(sparse=False)

train_dict = df_train[categorical_features + numerical_features].to_dict(orient="records")
X_train = dv.fit_transform(train_dict)
X_train.shape
```

```python
val_dict = df_val[categorical_features + numerical_features].to_dict(orient="records")
X_val = dv.transform(val_dict)
X_val.shape
```

## Logistic regression: Binary classification

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(solver="liblinear", random_state=1)
model.fit(X_train, y_train)
```

```python
model.intercept_[0]
```

```python
model.coef_[0].round(3)
```

## Predictions

```python
y_pred = model.predict_proba(X_val)[:,1]
y_pred
```

```python
churn_decision = (y_pred >= 0.5)
churn_decision
```

## Accuracy

```python
(y_val == churn_decision).mean()
```

About 80% of the cases the model predicted churn or no churn correctly

```python
df_pred = pd.DataFrame()
df_pred["probability"] = y_pred
df_pred["prediction"] = churn_decision.astype(int)
df_pred["actual"] = y_val

df_pred["correct"] = (df_pred["prediction"] == df_pred["actual"])
df_pred["correct"].mean()
```

## Final model

```python
dicts_full_train = df_full_train[categorical_features + numerical_features].to_dict(orient="records")

dvf = DictVectorizer(sparse=False)
X_full_train = dvf.fit_transform(dicts_full_train)

y_full_train = df_full_train["churn"].values
```

```python
model = LogisticRegression(solver="liblinear", random_state=1)
model.fit(X_full_train, y_full_train)
```

```python
# Check test set

dicts_test = df_test[categorical_features + numerical_features].to_dict(orient="records")
X_test = dvf.transform(dicts_test)

y_pred = model.predict_proba(X_test)[:,1]
churn_decision = (y_pred >= 0.5)
(churn_decision == y_test).mean()
```

Accuracy on the test set is about 81.26%. On validation we had 80.3%. The two numbers are close, which is what we want to see: the model performs on unseen data about as well as on the data it was tuned with. A large gap between the two would be a sign of overfitting.
