```python
u = np.array(
    [[3, 1, 2],
     [4, 5, 7],
     [9, 6, 8]]
)
v = np.array(
    [[9, 1, 3],
     [2, 6, 7],
     [8, 4, 5]]
)
```

```python
np.matmul(u,v)
```

```python
I = np.eye(3)
I
```

```python
w = np.matmul(u,v)
w
```

```python
np.matmul(w,I)
```

Let's demonstrate the following

$$
X X^{-1} = X^{-1} X = I
$$

```python
np.matmul(w, np.linalg.inv(w)).round()
```

```python
np.matmul(np.linalg.inv(w), w).round()
```

$$
(AB)(B^{-1} A^{-1}) = (B^{-1} A^{-1})(AB) = I
$$

```python
w = np.matmul(u, v)
w
```

```python
temp = np.matmul(np.linalg.inv(v), np.linalg.inv(u))
temp
```

```python
np.matmul(w, temp).round()
```

```python
np.matmul(temp, w).round()
```
